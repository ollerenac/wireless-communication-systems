### P1 — Ecuación de Friis

Un enlace inalámbrico opera a $f_c = 2.4$ GHz. La potencia de transmisión es $P_t = 20$ dBm y ambas antenas son isotrópicas ($G_t = G_r = 0$ dBi). La distancia entre transmisor y receptor es $d = 500$ m.

**(a)** Calcule la longitud de onda $\lambda$.

**(b)** Calcule la potencia recibida $P_r$ en dBm usando la ecuación de Friis en espacio libre.

**(c)** Si el umbral de sensibilidad del receptor es $-90$ dBm, ¿existe margen de enlace suficiente? ¿De cuántos dB?

**Solución P1 — Ecuación de Friis**  
*(Sesión 01, §1 — Pérdida de propagación en espacio libre)*

**(a)** Longitud de onda:
$$\lambda = \frac{c}{f_c} = \frac{3\times10^8}{2.4\times10^9} = 0.125\text{ m}$$

**(b)** Pérdida en espacio libre (Friis):
$$\mathrm{PL}_{\mathrm{FS}} = 20\log_{10}\!\left(\frac{4\pi d}{\lambda}\right) = 20\log_{10}\!\left(\frac{4\pi \times 500}{0.125}\right) = 20\log_{10}(50265) = 94.03\text{ dB}$$

Potencia recibida:
$$P_r = P_t + G_t + G_r - \mathrm{PL}_{\mathrm{FS}} = 20 + 0 + 0 - 94.03 = \mathbf{-74.0\text{ dBm}}$$

**(c)** Margen de enlace:
$$M = P_r - P_{\mathrm{sen}} = -74.0 - (-90) = \mathbf{+16\text{ dB}} \quad \checkmark$$

El enlace es viable con ~16 dB de margen de shadowing disponible.

> **Concepto clave:** La Friis muestra que $P_r \propto d^{-2}$ en espacio libre — cada vez que la distancia se duplica, la potencia recibida cae 6 dB.

### P3 — Shadowing

**(a)** Explica qué fenómeno físico causa el efecto de *shadowing* y por qué se modela como una variable aleatoria log-normal $X_\sigma \sim \mathcal{N}(0, \sigma^2)$ en escala dB.

**(b)** ¿Qué representa el parámetro $\sigma$ (desviación estándar del shadowing)? Da valores típicos para entornos de interior y exterior urbano.

**(c)** Si $\sigma = 8$ dB, ¿cuál es la probabilidad de que la pérdida de propagación real supere la media predicha por el modelo log-distancia en más de 10 dB?

**Solución P3 — Shadowing**  
*(Sesión 01, §3 — Shadowing)*

**(a)** El shadowing ocurre por **obstáculos físicos** (edificios, montañas, vegetación) que bloquean la señal. En escala logarítmica, el efecto total de múltiples obstáculos independientes es una **suma** de variables aleatorias → por el Teorema Central del Límite, esta suma converge a una distribución Gaussiana. Al expresar la pérdida en dB, una atenuación multiplicativa en escala lineal se vuelve aditiva en dB, justificando $X_\sigma \sim \mathcal{N}(0,\sigma^2)$ en dB.

**(b)** $\sigma$ es la **desviación estándar de la variabilidad de propagación** en dB:
- Interior (office): $\sigma \approx 5\text{–}8$ dB  
- Exterior urbano: $\sigma \approx 8\text{–}12$ dB  
- Suburbano: $\sigma \approx 4\text{–}8$ dB

**(c)** $P(X_\sigma > 10\text{ dB}) = Q\!\left(\frac{10}{8}\right) = Q(1.25) \approx \mathbf{10.6\%}$

Esto significa que en el 10.6% de las ubicaciones a igual distancia, la pérdida real supera en más de 10 dB la predicción del modelo determinista.

### P4 — Delay Spread y Mean Excess Delay

Un canal multipercusión tiene tres trayectorias con las siguientes características:

| Trayectoria | $|a_i|^2$ | $\tau_i$ |
|:-----------:|:---------:|:---------:|
| 0 | 0.6 | 0 ns |
| 1 | 0.3 | 120 ns |
| 2 | 0.1 | 300 ns |

**(a)** Calcule el *mean excess delay* $\bar{\tau}$.

**(b)** Calcule el *RMS delay spread* $\sigma_\tau$.

**Solución P4 — Delay Spread**  
*(Sesión 01, §4 — Propagación multipercusión)*

**(a)** Mean excess delay:
$$\bar{\tau} = \frac{\sum_i |a_i|^2 \tau_i}{\sum_i |a_i|^2} = \frac{0.6\cdot0 + 0.3\cdot120 + 0.1\cdot300}{1.0} = \frac{0 + 36 + 30}{1} = \mathbf{66\text{ ns}}$$

$$\overline{\tau^2} = 0.6\cdot0^2 + 0.3\cdot120^2 + 0.1\cdot300^2 = 0 + 4320 + 9000 = 13320\text{ ns}^2$$
$$\sigma_\tau = \sqrt{\overline{\tau^2} - \bar{\tau}^2} = \sqrt{13320 - 66^2} = \sqrt{13320 - 4356} = \sqrt{8964} \approx \mathbf{94.7\text{ ns}}$$

### P8 — Distribución de Rayleigh

**(a)** Explica la condición física que justifica que la envolvente $r = |h|$ siga una distribución de Rayleigh en un canal NLOS. ¿Qué supuesto estadístico sobre los ecos individuales se requiere?

**(b)** Si $h = h_I + j h_Q$ con $h_I, h_Q \sim \mathcal{N}(0, \sigma^2)$ i.i.d., escribe la función densidad de probabilidad (PDF) de la envolvente $r = |h|$.

**(c)** ¿Qué distribución sigue la SNR instantánea $\gamma = r^2 \bar{\gamma} / (2\sigma^2)$ y cuál es su media $\bar{\gamma}$?

**Solución P8 — Distribución de Rayleigh**  
*(Sesión 01, §7 — Rayleigh fading)*

**(a)** En entorno NLOS, la señal es la suma de $N\gg1$ ecos con amplitudes y fases **independientes** y distribuidas uniformemente: $h = \sum_i a_i e^{j\phi_i}$. Por el **Teorema Central del Límite**, las componentes I y Q convergen a gaussianas i.i.d.: $h_I, h_Q \sim \mathcal{N}(0,\sigma^2)$ → la envolvente $r=|h|$ sigue una distribución de **Rayleigh**.

**(b)**
$$f_R(r) = \frac{r}{\sigma^2}\exp\!\left(-\frac{r^2}{2\sigma^2}\right), \quad r \geq 0$$

**(c)** La potencia instantánea $p = r^2$ sigue una distribución exponencial. La SNR instantánea $\gamma = r^2\bar{\gamma}/(2\sigma^2)$ satisface:
$$f_\gamma(\gamma) = \frac{1}{\bar{\gamma}}\exp\!\left(-\frac{\gamma}{\bar{\gamma}}\right), \quad \gamma \geq 0$$
Es decir, $\gamma$ es **exponencialmente distribuida** con media $\bar{\gamma}$. Esta distribución tiene una cola larga que modela los *deep fades*.

### P9 — Canal Rician

**(a)** ¿En qué escenario físico es apropiado usar el modelo Rician en lugar del modelo Rayleigh?

**(b)** Define el factor $K$ de Rician. ¿Qué ocurre en los límites $K \to 0$ y $K \to \infty$?

**(c)** Para un factor $K = 10$ dB, ¿la variabilidad del canal es mayor o menor que en Rayleigh? Justifica.

**Solución P9 — Canal Rician**  
*(Sesión 01, §8 — Rician fading)*

**(a)** El modelo Rician aplica cuando existe una **componente LOS (Line-of-Sight)** determinista además de los ecos difusos. Ejemplos: enlace satélite, interior de oficina con visión directa, picoceldas.

**(b)** El factor K es la razón entre la potencia de la componente LOS y la potencia de las componentes difusas:
$$K = \frac{A^2/2}{\sigma^2}$$
donde $A$ es la amplitud del camino LOS.  
- $K\to0$: sin componente LOS → **Rayleigh** (caso más severo de fading).  
- $K\to\infty$: componente LOS dominante → canal **AWGN** (determinista).

**(c)** $K=10$ dB $= 10$ → el 10/11 ≈ 91% de la potencia es LOS y solo el 9% difuso. La variabilidad del canal es **mucho menor** que en Rayleigh: el canal es casi determinista. La BER se acerca a la del canal AWGN.

### P11 — Detector de Máxima Verosimilitud (ML)

**(a)** Escribe el criterio de decisión del detector ML: $\hat{m} = \arg\min_m \, d(y, s_m)$. ¿Bajo qué supuesto sobre el ruido el detector ML es equivalente al detector de mínima distancia euclídea?

**(b)** Para BPSK con señales $s_0 = -\sqrt{E_b}$ y $s_1 = +\sqrt{E_b}$, ¿cuál es la frontera de decisión y la región de decisión para cada símbolo?

**(c)** ¿Cómo cambian las regiones de decisión si las señales no son equiprobables ($P(s_0) \neq P(s_1)$)?

**Solución P11 — Detector ML**  
*(Sesión 02, §1 — Del canal a la decisión)*

**(a)** Criterio ML: $\hat{m} = \arg\max_m p(y\mid s_m)$. Bajo ruido AWGN ($n\sim\mathcal{N}(0,N_0/2)$), maximizar la verosimilitud es equivalente a minimizar $\|y-s_m\|^2$ (distancia euclídea), pues $p(y\mid s_m)\propto\exp(-\|y-s_m\|^2/N_0)$. Si el ruido no fuera gaussiano (ej. impulsivo), el detector ML diferiría del de mínima distancia.

**(b)** Para BPSK con $s_0=-\sqrt{E_b}$ y $s_1=+\sqrt{E_b}$:
- **Frontera de decisión:** $y=0$ (equidistante de ambas señales)
- **Región $\hat{m}=0$:** $y<0$  
- **Región $\hat{m}=1$:** $y>0$

**(c)** Si $P(s_0)>P(s_1)$, el detector MAP desplaza la frontera hacia $s_1$: el umbral se mueve a $y^*=\frac{N_0}{4\sqrt{E_b}}\ln\frac{P(s_0)}{P(s_1)}>0$. En la práctica, los símbolos suelen asumirse equiprobables, haciendo MAP ≡ ML.

### P13 — BER de BPSK

**(a)** A partir de la geometría del espacio de señales, justifica que la probabilidad de error de bit de BPSK es $P_b = Q\!\left(\sqrt{2E_b/N_0}\right)$. Indica claramente el papel de la distancia mínima $d_{\min}$ y de la varianza del ruido.

**(b)** Calcula $P_b^{\mathrm{BPSK}}$ para $E_b/N_0 = 6$ dB. (Dato: $Q(2.0) \approx 0.023$, $Q(3.0) \approx 0.0013$, $Q(3.46) \approx 2.7\times10^{-4}$.)

**(c)** ¿Cuánto $E_b/N_0$ se necesita para alcanzar $P_b = 10^{-5}$? (Dato: $Q^{-1}(10^{-5}) \approx 4.42$.)

**Solución P13 — BER de BPSK**  
*(Sesión 02, §2.2 — Derivación de la BER)*

**(a)** Bajo $s_1=+\sqrt{E_b}$ y ruido $n\sim\mathcal{N}(0,N_0/2)$, el error ocurre cuando $y = \sqrt{E_b}+n < 0$, es decir $n < -\sqrt{E_b}$:
$$P_b = P(n < -\sqrt{E_b}) = Q\!\left(\frac{\sqrt{E_b}}{\sqrt{N_0/2}}\right) = Q\!\left(\sqrt{\frac{2E_b}{N_0}}\right)$$

La distancia mínima entre señales es $d_{\min}=2\sqrt{E_b}$, y el argumento de $Q$ es $d_{\min}/(2\sqrt{N_0/2}) = \sqrt{2E_b/N_0}$.

**(b)** $E_b/N_0 = 6\text{ dB} = 3.98$. Argumento: $\sqrt{2\times3.98} = \sqrt{7.96} = 2.82$.

$$P_b = Q(2.82) \approx \mathbf{2.4\times10^{-3}}$$

(Interpolando entre $Q(2.0)=0.023$ y $Q(3.0)=0.0013$.)

**(c)** Para $P_b = 10^{-5}$: $Q^{-1}(10^{-5})=4.42$, entonces $\sqrt{2E_b/N_0}=4.42$ → $E_b/N_0=4.42^2/2=9.77$ lineal $=\mathbf{9.9\text{ dB}}$.

### P14 — QPSK y Eficiencia Espectral

**(a)** QPSK usa 4 símbolos, cada uno codificando 2 bits. Muestra que su BER por bit satisface $P_b^{\mathrm{QPSK}} = Q(\sqrt{2E_b/N_0})$, la misma que BPSK.

**(b)** Si la BER es igual a BPSK, ¿qué ventaja ofrece QPSK?

**(c)** Explica el concepto de ortogonalidad entre las componentes I y Q en QPSK y cómo permite detectar cada componente de forma independiente.

**Solución P14 — QPSK**  
*(Sesión 02, §3 — QPSK)*

**(a)** En QPSK, la energía total por símbolo es $E_s=2E_b$. Las componentes I y Q son independientes y cada una equivale a BPSK con energía $E_b$ por bit: $P_{b,I}=Q(\sqrt{2E_b/N_0})$. Por simetría $P_{b,Q}=P_{b,I}$, y dado que I y Q no interfieren, la BER total por bit es:
$$P_b^{\mathrm{QPSK}} = Q\!\left(\sqrt{\frac{2E_b}{N_0}}\right) \quad \text{(igual que BPSK)}$$

**(b)** QPSK transmite **2 bits por símbolo**. A igual $E_b/N_0$ y misma BER que BPSK, QPSK requiere solo la **mitad del ancho de banda** (o bien, a igual ancho de banda, duplica la tasa de datos). Esta es la ventaja clave de QPSK.

**(c)** Las portadoras $\cos(2\pi f_c t)$ y $\sin(2\pi f_c t)$ son ortogonales: $\int_0^{T_s}\cos\cdot\sin\,dt=0$. El receptor puede hacer la correlación con cada portadora de forma **independiente**, recuperando la componente I sin contaminación de Q y viceversa. Esto permite un receptor de dos ramas separadas, sin interferencia cruzada.

### P15 — Penalización de SNR en M-QAM

La BER aproximada de M-QAM es:
$$P_b^{M\text{-QAM}} \approx \frac{4}{\log_2 M}\left(1 - \frac{1}{\sqrt{M}}\right) Q\!\left(\sqrt{\frac{3\log_2 M}{M-1}\,\frac{E_b}{N_0}}\right)$$

**(a)** Para 16-QAM ($M=16$) y 64-QAM ($M=64$), calcula el argumento de la función $Q$ en función de $E_b/N_0$ lineal.

**(b)** Estima la penalización de SNR en dB de 16-QAM y 64-QAM respecto a BPSK para alcanzar la misma BER.

**Solución P15 — Penalización de SNR en M-QAM**  
*(Sesión 02, §4 — Tabla de penalización)*

El argumento de $Q$ en M-QAM es $\sqrt{\dfrac{3\log_2 M}{M-1}\cdot\dfrac{E_b}{N_0}}$.

**(a)**
- **16-QAM** ($M=16$): factor $= \dfrac{3\times4}{15} = 0.800$ → argumento $= \sqrt{0.800\,\gamma_b}$  
- **64-QAM** ($M=64$): factor $= \dfrac{3\times6}{63} = 0.286$ → argumento $= \sqrt{0.286\,\gamma_b}$

**(b)** Para la misma BER, el argumento de $Q$ debe ser igual al de BPSK ($\sqrt{2\gamma_b}$):
$$\frac{3\log_2 M}{M-1}\gamma_{b,M} = 2\gamma_{b,\mathrm{BPSK}} \implies \text{penalización} = 10\log_{10}\!\frac{2(M-1)}{3\log_2 M}$$

| Modulación | Penalización vs BPSK |
|:---:|:---:|
| 16-QAM | $10\log_{10}(2.5) = \mathbf{4.0\text{ dB}}$ |
| 64-QAM | $10\log_{10}(7.0) = \mathbf{8.5\text{ dB}}$ |

> **Interpretación:** Cada aumento en el orden de la constelación cuesta $\sim$4–5 dB adicionales de SNR para mantener la misma BER.

```python
# Cálculos opcionales P15
import numpy as np
for M in [4, 16, 64, 256]:
    factor = 3 * np.log2(M) / (M - 1)
    penalty_dB = 10 * np.log10(1 / factor)
    print(f"M={M:4d}  SNR factor = {factor:.4f}  penalización vs BPSK = {penalty_dB:.2f} dB")
```

### P16 — Codificación Gray

**(a)** Explica por qué la codificación Gray reduce la BER en modulaciones de alto orden (M-QAM). ¿Qué propiedad debe cumplir el mapeo Gray entre bits y símbolos vecinos?

**(b)** Escribe el mapeo Gray para QPSK (4 símbolos, 2 bits). Verifica que símbolos adyacentes difieren en exactamente 1 bit.

**(c)** Sin Gray coding en 16-QAM, ¿cuántos bits en error esperarías por cada error de símbolo en promedio? ¿Y con Gray coding?

**Solución P16 — Codificación Gray**  
*(Sesión 02, §5 — Gray Coding)*

**(a)** Con Gray coding, **símbolos adyacentes en la constelación difieren en exactamente 1 bit**. Como la mayoría de los errores de símbolo producen una decisión al símbolo más cercano (vecino inmediato), con Gray coding casi todos los errores de símbolo causan solo **1 bit erróneo**. Sin Gray coding, un error de símbolo puede causar hasta $\log_2 M$ bits erróneos.

Resultado: $P_b \approx P_s/\log_2 M$ con Gray (vs. $P_b \approx P_s/2$ sin Gray para M grande).

**(b)** Mapeo Gray para QPSK (los 4 puntos del plano IQ, en orden antihorario):
$$(-1,-1)\to\mathbf{00},\quad(-1,+1)\to\mathbf{01},\quad(+1,+1)\to\mathbf{11},\quad(+1,-1)\to\mathbf{10}$$
Verificación: puntos adyacentes 00↔01 difieren en 1 bit ✓, 01↔11 difieren en 1 bit ✓, etc.

**(c)**  
- **Sin Gray (binario natural) en 16-QAM:** un error de símbolo puede cambiar hasta 4 bits → promedio $\approx 2$ bits por error de símbolo.  
- **Con Gray coding:** un error de símbolo al vecino más cercano → **1 bit erróneo** en promedio. Mejora de $\sim$3 dB en BER efectiva a igual $P_s$.

### P19 — BER de BPSK en Canal Rayleigh

La BER de BPSK en canal Rayleigh se obtiene promediando la BER condicional sobre la distribución exponencial de la SNR:
$$P_b = \int_0^\infty Q(\sqrt{2\gamma})\,\frac{1}{\bar{\gamma}}e^{-\gamma/\bar{\gamma}}\,d\gamma = \frac{1}{2}\left(1 - \sqrt{\frac{\bar{\gamma}}{1+\bar{\gamma}}}\right) \approx \frac{1}{4\bar{\gamma}} \quad (\bar{\gamma} \gg 1)$$

**(a)** Calcula $P_b$ en Rayleigh para $\bar{\gamma} = 10$ dB y $\bar{\gamma} = 20$ dB usando la aproximación $1/(4\bar{\gamma})$.

**(b)** Para el mismo $\bar{\gamma} = 10$ dB, calcula $P_b$ de BPSK en AWGN. ¿Cuántos órdenes de magnitud separa ambos resultados?

**(c)** ¿Por qué la curva BER en Rayleigh tiene pendiente $1/\bar{\gamma}$ (lineal en escala log-log) y no exponencialmente decreciente como en AWGN?

**Solución P19 — BER de BPSK en Canal Rayleigh**  
*(Sesión 01, §6–§7 + Sesión 02, §2.2)*

**(a)**  
- $\bar{\gamma}=10$ dB $=10$: $P_b \approx 1/(4\times10) = \mathbf{2.5\times10^{-2}}$  
- $\bar{\gamma}=20$ dB $=100$: $P_b \approx 1/(4\times100) = \mathbf{2.5\times10^{-3}}$

**(b)** Para $\bar{\gamma}=10$ dB en AWGN: $P_b=Q(\sqrt{20})=Q(4.47)\approx3.9\times10^{-6}$.

Diferencia: $2.5\times10^{-2}$ vs $3.9\times10^{-6}$ → **más de 4 órdenes de magnitud** peor en Rayleigh.

**(c)** La SNR instantánea $\gamma$ tiene distribución exponencial $f(\gamma)=\frac{1}{\bar{\gamma}}e^{-\gamma/\bar{\gamma}}$, que asigna probabilidad positiva a valores $\gamma\approx0$. En estos *deep fades*, la BER condicional $Q(\sqrt{2\gamma})\approx1/2$. Al promediar $P_b=\int_0^\infty Q(\sqrt{2\gamma})\frac{e^{-\gamma/\bar{\gamma}}}{\bar{\gamma}}d\gamma$, los instantes de $\gamma\approx0$ dominan el resultado, produciendo un decaimiento polinomial $1/(4\bar{\gamma})$ en lugar del exponencial del canal AWGN.

```python
# Cálculos opcionales P19
from scipy.special import erfc
import numpy as np

def Q(x):
    return 0.5 * erfc(x / np.sqrt(2))

for snr_db in [10, 20]:
    g = 10**(snr_db/10)
    ber_rayleigh_approx = 1 / (4 * g)
    ber_awgn = Q(np.sqrt(2 * g))
    print(f"SNR={snr_db} dB | BER Rayleigh≈{ber_rayleigh_approx:.2e} | BER AWGN={ber_awgn:.2e}")
```

### P20 — Espacio de Señales y Ortogonalidad I/Q

**(a)** Explica el concepto de espacio de señales (*signal space*). ¿Qué representan los ejes del espacio de señales para una modulación QAM?

**(b)** Las portadoras $\cos(2\pi f_c t)$ y $\sin(2\pi f_c t)$ son ortogonales en $[0, T_s]$. Demuestra esta ortogonalidad calculando su producto interno.

**(c)** ¿Qué ventaja práctica de implementación ofrece la ortogonalidad I/Q en un receptor QAM?

**Solución P20 — Espacio de Señales y Ortogonalidad I/Q**  
*(Sesión 02, §1 — Espacio de señales)*

**(a)** El espacio de señales es un espacio vectorial donde cada señal $s_m(t)$ se representa como un punto (vector) con coordenadas $(s_{m,I}, s_{m,Q})$. Los ejes son las funciones base ortogonales normalizadas: eje I = $\sqrt{2/T_s}\cos(2\pi f_c t)$ y eje Q = $\sqrt{2/T_s}\sin(2\pi f_c t)$. Para QAM, los ejes I y Q corresponden a la amplitud en fase y en cuadratura.

**(b)** Producto interno de las portadoras (para $f_c T_s \in \mathbb{Z}$, i.e., número entero de ciclos en $T_s$):
$$\int_0^{T_s}\cos(2\pi f_c t)\sin(2\pi f_c t)\,dt = \frac{1}{2}\int_0^{T_s}\sin(4\pi f_c t)\,dt = \left[-\frac{\cos(4\pi f_c t)}{8\pi f_c}\right]_0^{T_s} = 0 \quad \checkmark$$

**(c)** La ortogonalidad I/Q permite que el receptor tenga **dos ramas independientes**: el mezclador I ($\times\cos$) recupera solo la componente I sin contaminación de Q, y viceversa. Esto permite detectar una constelación 2D (como 16-QAM) con dos detectores 1D independientes, reduciendo la complejidad y evitando la necesidad de un ecualizador cruzado.

### P21 — Motivación de OFDM

**(a)** Explica por qué transmitir a alta tasa con una sola portadora en un canal multipercusión (con $\sigma_\tau > 0$) genera interferencia entre símbolos (ISI). Ilustra con un ejemplo numérico: $\sigma_\tau = 5\,\mu$s, $T_s = 1\,\mu$s.

**(b)** ¿Cómo resuelve OFDM el problema de ISI de forma elegante? Explica el principio fundamental sin entrar en el CP todavía.

**(c)** ¿Cuál es la relación entre el número de subportadoras $N$ y el ancho de banda de subportadora $\Delta f$?

**Solución P21 — Motivación de OFDM**  
*(Sesión 03, §1 — De la portadora única al multiportadora)*

**(a)** Con portadora única a tasa $R=1/T_s$, el símbolo dura $T_s$. Si $\sigma_\tau > T_s$, la energía de un símbolo se derrama sobre símbolos siguientes durante $\sigma_\tau$ (la respuesta impulsional del canal), contaminándolos.

Ejemplo numérico: $\sigma_\tau=5\,\mu$s, $T_s=1\,\mu$s → la "cola" del canal dura $\approx5T_s$: el símbolo $k$ interfiere con los símbolos $k+1,\ldots,k+5$ → **ISI severa**.

**(b)** OFDM divide el ancho de banda $B$ en $N$ subportadoras angostas, cada una con $\Delta f = B/N$. El símbolo OFDM dura $T_s=1/\Delta f = N/B$, que se hace **mucho mayor** que $\sigma_\tau$ al elegir $N$ suficientemente grande. Cada subportadora experimenta entonces fading plano (flat fading) → sin ISI dentro de cada subportadora. El canal multipercusión se convierte en $N$ canales escalares independientes.

**(c)** $N = B/\Delta f$ (número de subportadoras = cociente entre el ancho de banda total y el espaciado entre subportadoras). El producto $N\Delta f = B$ es el ancho de banda total.

### P22 — Ortogonalidad de Subportadoras

Las subportadoras OFDM son $\phi_k(t) = e^{j2\pi k \Delta f \, t}$, con $\Delta f = 1/T_s$.

**(a)** Demuestra que $\frac{1}{T_s}\int_0^{T_s} \phi_k(t)\,\phi_l^*(t)\,dt = \delta_{kl}$ (delta de Kronecker).

**(b)** En el dominio discreto, demuestra que $\frac{1}{N}\sum_{n=0}^{N-1} e^{j2\pi(k-l)n/N} = \delta_{kl}$.

**(c)** ¿Qué ocurre si se introduce un error de frecuencia $\epsilon \Delta f$ (con $\epsilon \neq 0$) en el oscilador del receptor? ¿Qué fenómeno genera?

**Solución P22 — Ortogonalidad de Subportadoras**  
*(Sesión 03, §2 — Ortogonalidad y la DFT)*

**(a)** Para $k\neq l$, con $\Delta f=1/T_s$:
$$\frac{1}{T_s}\int_0^{T_s} e^{j2\pi k\Delta f t}\cdot e^{-j2\pi l\Delta f t}\,dt = \frac{1}{T_s}\int_0^{T_s} e^{j2\pi(k-l)t/T_s}\,dt = \frac{e^{j2\pi(k-l)}-1}{j2\pi(k-l)} = \frac{0}{j2\pi(k-l)} = 0 \; \checkmark$$
Para $k=l$: el integrando es 1 → resultado $= 1$. $\therefore \delta_{kl}$.

**(b)** $\frac{1}{N}\sum_{n=0}^{N-1}e^{j2\pi(k-l)n/N}$: suma geométrica con razón $r=e^{j2\pi(k-l)/N}$.  
- $k=l$: suma de $N$ unos dividida entre $N$ $= 1$.  
- $k\neq l$: $= \frac{1}{N}\cdot\frac{1-r^N}{1-r} = \frac{1-e^{j2\pi(k-l)}}{N(1-r)} = \frac{0}{N(1-r)} = 0$ $\checkmark$

**(c)** Un error de frecuencia (CFO) $\epsilon\Delta f$ rompe la ortogonalidad: el término $e^{j2\pi(k-l+\epsilon)n/N}$ ya no suma cero para $k\neq l$ → genera **ICI** (*Inter-Carrier Interference*): cada subportadora contamina a todas las demás, degradando la BER.

### P23 — Rol de IFFT y FFT en OFDM

**(a)** Explica por qué la señal OFDM en tiempo discreto $x[n] = \frac{1}{\sqrt{N}}\sum_{k=0}^{N-1} X[k]\,e^{j2\pi kn/N}$ corresponde exactamente a la IFFT del vector de símbolos $\{X[k]\}$.

**(b)** En el receptor, tras el CP y el canal, se aplica la FFT. ¿Qué señal se obtiene en cada subportadora $k$? Escribe la expresión.

**(c)** ¿Por qué las subportadoras en los extremos del espectro suelen dejarse nulas (guardabandas) en implementaciones prácticas?

**Solución P23 — Rol de IFFT y FFT**  
*(Sesión 03, §2 y §4.2 — Cadena OFDM)*

**(a)** La señal OFDM: $x[n] = \frac{1}{\sqrt{N}}\sum_{k=0}^{N-1}X[k]\,e^{j2\pi kn/N}$ es exactamente la **IDFT** de $\{X[k]\}$ (normalizada). La IFFT es el algoritmo rápido para computar la IDFT en $O(N\log N)$ en lugar de $O(N^2)$. El transmisor simplemente aplica IFFT al vector de símbolos QAM en frecuencia para obtener la señal en tiempo.

**(b)** Tras eliminar el CP y aplicar FFT (con CP convirtiendo la convolución en circular):
$$Y[k] = H[k]\,X[k] + W[k]$$
Cada subportadora ve solo una **multiplicación escalar** por la respuesta en frecuencia del canal $H[k]$ (más ruido aditivo). El canal matricial disperso se diagonaliza.

**(c)** Las subportadoras en los extremos del espectro se dejan nulas (**guardabandas**) por dos razones:  
1. Los filtros de RF no son ideales → las subportadoras del borde "sangran" hacia bandas adyacentes.  
2. Los efectos de aliasing en la IFFT/FFT son mayores en los extremos.  
En 5G NR, tipicamente se reservan ~10% de subportadoras como guardabandas.

### P24 — Prefijo Cíclico

**(a)** Define el prefijo cíclico (CP) y explica cómo se construye a partir de las últimas muestras del símbolo OFDM.

**(b)** ¿Cuál es la longitud mínima del CP en muestras para un canal con retardo máximo de $L-1$ muestras? ¿Qué ocurre si el CP es demasiado corto?

**(c)** El CP introduce un *overhead*. Calcula el overhead en porcentaje si $N = 512$ y $N_{CP} = 64$.

**Solución P24 — Prefijo Cíclico**  
*(Sesión 03, §3 — El prefijo cíclico)*

**(a)** El CP es una **copia de las últimas $N_{CP}$ muestras** del símbolo OFDM, que se antepone al inicio:
$$x_{\mathrm{CP}} = [\underbrace{x[N-N_{CP}],\ldots,x[N-1]}_{\text{CP}},\, x[0],\ldots,x[N-1]]$$
El símbolo transmitido tiene duración total $N+N_{CP}$ muestras.

**(b)** La longitud mínima es $N_{CP} \geq L-1$ muestras, donde $L$ es el número de taps del canal (retardo máximo = $L-1$ muestras). Si $N_{CP} < L-1$, la cola del canal del símbolo anterior invade las muestras útiles del símbolo actual → **ISI** y pérdida de la ortogonalidad.

**(c)** Overhead $= \dfrac{N_{CP}}{N+N_{CP}} = \dfrac{64}{512+64} = \dfrac{64}{576} \approx \mathbf{11.1\%}$

de la tasa de transmisión se dedica al CP (bits "gastados" en protección, sin información).

### P25 — Convolución Lineal vs Circular

**(a)** La convolución del canal con el símbolo OFDM sin CP es *lineal*: $y[n] = \sum_{l=0}^{L-1} h[l]\,x[n-l]$. Explica por qué esto provoca ISI entre símbolos OFDM consecutivos.

**(b)** Al añadir el CP (copia de las últimas $N_{CP} \geq L-1$ muestras al inicio) y eliminarlo en recepción, la convolución efectiva se convierte en *circular*: $y[n] = \sum_{l=0}^{L-1} h[l]\,x[(n-l)\bmod N]$. Explica intuitivamente por qué.

**(c)** ¿Qué propiedad de la FFT convierte la convolución circular en una multiplicación puntual en frecuencia?

**Solución P25 — Convolución Lineal vs Circular**  
*(Sesión 03, §3–§4 — CP y ecualización)*

**(a)** Sin CP, la convolución lineal del canal $y[n]=\sum_{l=0}^{L-1}h[l]\,x[n-l]$ al inicio del símbolo (para $n<L-1$) requiere muestras del símbolo **anterior**: $x[-1], x[-2],\ldots$ → el símbolo anterior contamina el actual durante $L-1$ muestras → **ISI entre símbolos OFDM**.

**(b)** Con CP de longitud $\geq L-1$: las muestras $x[-1], x[-2],\ldots, x[-(L-1)]$ requeridas por el canal son exactamente la copia del final del símbolo actual ($x[N-1], x[N-2],\ldots$), que es lo que contiene el CP. Al eliminar el CP en recepción, la convolución se cierra sobre sí misma en módulo $N$: $y[n]=\sum h[l]\,x[(n-l)\bmod N]$.

**(c)** **Teorema de convolución circular de la DFT:** $\mathrm{DFT}\{x\circledast h\} = \mathrm{DFT}\{x\}\cdot\mathrm{DFT}\{h\}$ (producto puntual en frecuencia). Por esto, $Y[k]=H[k]X[k]$: el canal multitap se convierte en $N$ canales de 1 tap independientes.

### P26 — Ecualización de Un Tap

**(a)** Tras la FFT, la relación entre símbolo transmitido y recibido en la subportadora $k$ es: $Y[k] = H[k]\,X[k] + W[k]$. ¿Cuántos coeficientes necesita el ecualizador por subportadora?

**(b)** Escribe la operación del ecualizador Zero-Forcing (ZF) de un tap. ¿Qué condición sobre $H[k]$ hace que el ecualizador ZF sea inestable?

**(c)** Compara la complejidad de ecualización en OFDM con la de un ecualizador FIR en el dominio del tiempo para un canal de $L=100$ taps y $N=1024$ subportadoras.

**Solución P26 — Ecualización de Un Tap**  
*(Sesión 03, §4.5 — Ecualización de un tap)*

**(a)** Solo **1 coeficiente complejo** por subportadora: la división $\hat{X}[k]=Y[k]/H[k]$.

**(b)** Ecualizador ZF de un tap:
$$\hat{X}[k] = \frac{Y[k]}{H[k]} = X[k] + \frac{W[k]}{H[k]}$$
Inestable cuando $H[k]\approx0$ (nulo de frecuencia del canal, *deep fade en frecuencia*): la división amplifica el ruido $W[k]$ por $1/|H[k]|\to\infty$. En ese caso se prefiere un ecualizador MMSE: $\hat{X}[k]=H^*[k]/(|H[k]|^2+N_0)$.

**(c)** Comparación de complejidad por símbolo OFDM ($L=100$, $N=1024$):
- **Tiempo (FIR):** $L\times N = 100\times1024 = 102\,400$ multiplicaciones.  
- **OFDM (FFT+1-tap):** $N\log_2N + N = 1024\times10 + 1024 = 11\,264$ operaciones.  
OFDM es $\sim$9× más eficiente en este ejemplo, y la ventaja crece con $L$.

### P27 — ISI vs ICI: Condiciones de Diseño

Completa y justifica:

**(a)** Para evitar ISI entre símbolos OFDM: la longitud del CP debe satisfacer _______ en función del canal.

**(b)** Para evitar ICI (interferencia entre subportadoras): el espacio $\Delta f$ debe satisfacer _______ en relación con la movilidad del canal.

**(c)** Hay una tensión de diseño entre los requisitos (a) y (b). Explica en qué sentido son contrapuestos y cómo se resuelve esta tensión en 5G NR (pista: *numerology*).

**Solución P27 — ISI vs ICI: Condiciones de Diseño**  
*(Sesión 03, §3 y §4 — Diseño OFDM)*

**(a)** Para evitar ISI: $N_{CP} \geq L-1$ (en muestras), equivalentemente $T_{CP} \geq \tau_{\max}$ en tiempo continuo, donde $\tau_{\max}$ es el retardo máximo del canal (regla práctica: $\tau_{\max}\approx5\sigma_\tau$).

**(b)** Para evitar ICI: $\Delta f \gg f_{D,\max}$. Criterio práctico: $\Delta f \geq 10\,f_{D,\max}$, o equivalentemente $T_s = 1/\Delta f \ll T_c$.

**(c)** **Tensión de diseño:**  
- Reducir ISI pide $T_s$ grande (símbolo largo → más espacio para el CP sin overhead excesivo) → $\Delta f$ pequeño.  
- Reducir ICI pide $\Delta f$ grande → $T_s$ pequeño.  

**5G NR** resuelve esto con el concepto de *numerology* ($\mu$): $\Delta f = 15\times2^\mu$ kHz. Para alta movilidad (vehículo, mmWave) se usa $\mu=3$ ($\Delta f=120$ kHz, $T_s=8.3\,\mu$s). Para entornos estáticos de baja frecuencia se usa $\mu=0$ ($\Delta f=15$ kHz, $T_s=66.7\,\mu$s). El sistema elige la numerología según el escenario.

$$\frac{N_{CP}}{N+N_{CP}} = \frac{72}{1096} \approx \mathbf{6.57\%}$$

> **Nota:** El overhead de guardabandas (los $1024-900=124$ subportadoras no utilizadas) reduce adicionalmente la eficiencia: si se incluyeran las 1024 subportadoras, $\eta$ sería $\frac{1024\times6\times0.75}{1096}\approx4.21$ bits/s/Hz. **(c)** Métodos de interpolación entre pilotos (de menor a mayor complejidad):

### P32 — Límite Absoluto de Shannon en $E_b/N_0$

**(a)** A partir de la capacidad $C = B\log_2(1 + \mathrm{SNR})$ y la relación $\mathrm{SNR} = (E_b/N_0)\cdot(R/B)$, demuestra que para transmisión confiable se requiere:
$$\frac{E_b}{N_0} > \frac{2^{R/B} - 1}{R/B}$$

**(b)** Calcula el límite cuando la eficiencia espectral $\eta = R/B \to 0$. ¿Cuál es el valor en dB? ¿A qué se llama este valor?

**(c)** ¿Qué significa prácticamente que ningún código puede transmitir de forma confiable por debajo de $-1.59$ dB de $E_b/N_0$?

**Solución P32 — Límite Absoluto de Shannon en $E_b/N_0$**  
*(Sesión 05, §1 — Límite de Shannon)*

**(a)** Partiendo de $C=B\log_2(1+\mathrm{SNR})$ con $\mathrm{SNR}=(E_b/N_0)(R/B)$, imponiendo $R=C$:
$$\frac{R}{B} = \log_2\!\left(1+\frac{E_b}{N_0}\cdot\frac{R}{B}\right) \implies 2^{R/B} = 1 + \frac{E_b}{N_0}\cdot\frac{R/B}{1}$$
$$\boxed{\frac{E_b}{N_0} > \frac{2^{R/B}-1}{R/B}}$$

**(b)** Para $\eta=R/B\to0$, usando la regla de L'Hôpital (o la expansión $2^\eta=e^{\eta\ln2}\approx1+\eta\ln2$ para $\eta\to0$):
$$\lim_{\eta\to0}\frac{2^\eta-1}{\eta} = \ln 2 \approx 0.693 \implies \mathbf{10\log_{10}(0.693) = -1.59\text{ dB}}$$
Este es el **Límite de Shannon** (Shannon limit) — la mínima $E_b/N_0$ posible para cualquier comunicación confiable.

**(c)** Ningún sistema de comunicaciones, por muy sofisticado que sea su código, puede transmitir sin errores si $E_b/N_0 < -1.59$ dB. Los mejores códigos modernos (LDPC, Polar) llegan a $\sim0.1$ dB del límite en bloque largo. Es un **límite fundamental de la física** de la información, no del diseño.

**(a)** Escribe la ecuación de actualización del mensaje de nodo de **variable** $v$ al nodo de **verificación** $c$ en el algoritmo Sum-Product (BP):
$$\mu_{v \to c} = \lambda_v^{(0)} + \sum_{c' \in \mathcal{N}(v) \setminus c} \mu_{c' \to v}$$
Explica qué representa $\lambda_v^{(0)}$.

**(b)** Escribe la ecuación de actualización del mensaje de nodo de **verificación** $c$ al nodo de **variable** $v$:
$$\mu_{c \to v} = 2\,\mathrm{arctanh}\!\left(\prod_{v' \in \mathcal{N}(c) \setminus v} \tanh\!\left(\frac{\mu_{v' \to c}}{2}\right)\right)$$

**(c)** ¿Cuándo se detiene el algoritmo BP? ¿Qué ocurre si el grafo tiene ciclos cortos?

**Solución P35 — Belief Propagation: Mensajes V→C y C→V**  
*(Sesión 05, §3.2 — Belief Propagation)*

**(a)** Mensaje $v\to c$ (actualización de nodo de variable):
$$\mu_{v\to c} = \lambda_v^{(0)} + \sum_{c'\in\mathcal{N}(v)\setminus c}\mu_{c'\to v}$$
$\lambda_v^{(0)}$ es el **LLR del canal** del bit $v$: información inicial ("priori") proveniente de la observación $y_v$. El mensaje $\mu_{v\to c}$ es el LLR total de $v$ excluyendo la información que llegó de $c$ (para evitar realimentación de la misma información).

$$\mu_{c\to v} = 2\,\mathrm{arctanh}\!\left(\prod_{v'\in\mathcal{N}(c)\setminus v}\tanh\!\frac{\mu_{v'\to c}}{2}\right)$$
Este mensaje es la **información extrínseca** que el chequeo de paridad $c$ aporta sobre $v$, basándose en todos los demás bits del chequeo. La operación $\tanh$ convierte LLRs a probabilidades, el producto impone la condición de paridad, y $\mathrm{arctanh}$ regresa a LLR.

**(c)** El algoritmo BP se detiene cuando: (1) todos los chequeos de paridad se satisfacen ($\mathbf{H}\hat{\mathbf{x}}=\mathbf{0}$), o (2) se alcanza el número máximo de iteraciones. Los **ciclos cortos** (longitud 4) hacen que los mensajes pasen por los mismos nodos repetidamente, violando la independencia asumida por BP → los mensajes se correlacionan → convergencia lenta o a un punto fijo erróneo.

### P36 — Log-Likelihood Ratio (LLR)

**(a)** Define el LLR para un bit binario $b \in \{0, 1\}$ dado el canal:
$$\lambda = \log \frac{P(b=0 \mid y)}{P(b=1 \mid y)}$$
¿Qué indica el signo de $\lambda$? ¿Y el módulo $|\lambda|$?

**(b)** Para un canal AWGN con $y = s + n$, donde $s \in \{+1, -1\}$ y $n \sim \mathcal{N}(0, N_0/2)$, demuestra que el LLR es $\lambda = 2y/N_0$ (o equivalentemente $\lambda = 4y E_b / N_0$ para energía por bit unitaria).

**(c)** Calcula el LLR para $y = +0.6$, $N_0/2 = 0.25$ (es decir, $N_0 = 0.5$). ¿Cuál es la decisión hard del decodificador?

**Solución P36 — Log-Likelihood Ratio (LLR)**  
*(Sesión 05, §3.2 — LLR)*

**(a)** $\lambda = \log\dfrac{P(b=0|y)}{P(b=1|y)}$.  
- **Signo:** $\lambda>0$ → más probable $b=0$; $\lambda<0$ → más probable $b=1$.  
- **Módulo:** $|\lambda|$ mide la **fiabilidad** (confianza) de la decisión. $|\lambda|\gg1$: alta confianza; $|\lambda|\approx0$: máxima incertidumbre.

**(b)** Para el canal AWGN con $s\in\{+1,-1\}$ y $n\sim\mathcal{N}(0,N_0/2)$ [donde $N_0/2$ es la varianza del ruido]:
$$\lambda = \log\frac{\exp\!\left(-\frac{(y-1)^2}{N_0}\right)}{\exp\!\left(-\frac{(y+1)^2}{N_0}\right)} = \frac{(y+1)^2-(y-1)^2}{N_0} = \frac{4y}{N_0} = \frac{2y}{N_0/2}$$

Usando la relación $\sigma^2=N_0/2$: $\lambda=2y/\sigma^2$. La forma compacta del enunciado $\lambda=2y/N_0$ corresponde a la convención $N_0\equiv\sigma^2$ (varianza de ruido), equivalente a $4y/N_0^{\text{(físico)}}$.

**(c)** Con $y=+0.6$, $N_0/2=0.25$ ($\sigma^2=0.25$, $N_0^{\text{(físico)}}=0.5$):
$$\lambda = \frac{2\times0.6}{0.25} = 4.8 \quad (>0 \implies \hat{b}=0)$$

La decisión hard es $\hat{b}=\mathbf{0}$ (con bastante confianza, $|\lambda|=4.8\gg1$).

### P38 — Parámetro de Bhattacharyya

**(a)** Define el parámetro de Bhattacharyya $Z(W) \in [0,1]$ para un canal binario simétrico $W$. ¿Qué valor de $Z$ indica un canal perfecto y cuál indica un canal borrado?

**(b)** Explica cómo se usa $Z(W_N^{(i)})$ para construir un código Polar de dimensión $k$: ¿cuáles índices $i$ se eligen como posiciones de información y cuáles como bits congelados?

**(c)** ¿Por qué el parámetro de Bhattacharyya es una métrica de *cota superior* de la probabilidad de error, y no la probabilidad de error exacta?

**Solución P38 — Parámetro de Bhattacharyya**  
*(Sesión 05, §4.1 — Bhattacharyya)*

**(a)** Para un canal binario simétrico (BMS) $W: \{0,1\}\to\mathcal{Y}$:
$$Z(W) = \sum_{y\in\mathcal{Y}}\sqrt{W(y|0)\cdot W(y|1)} \in [0,1]$$

- $Z(W)=0$: canal **perfecto** (distribuciones $W(\cdot|0)$ y $W(\cdot|1)$ no se solapan → sin error posible).  
- $Z(W)=1$: canal completamente **borrado** (las distribuciones son idénticas → sin información).

**(b)** Para construir un código Polar $(N, k)$:
1. Calcular $Z(W_N^{(i)})$ para $i=0,\ldots,N-1$.
2. Ordenar de menor a mayor (menor $Z$ = canal más confiable).
3. Las $k$ posiciones con menor $Z$ → **bits de información**. Las $N-k$ restantes → **bits congelados** (a 0).

**(c)** El parámetro de Bhattacharyya es una **cota superior** de la probabilidad de error: $P_e^{(i)}\leq Z(W_N^{(i)})$. Es una cota laxa (especialmente para canales buenos), pero tiene la ventaja de ser **recursivamente computable** en $O(N\log N)$ sin necesidad de simulación Monte Carlo. La probabilidad de error real puede ser órdenes de magnitud menor que $Z$.

**(b)** ¿Qué ventaja de los códigos LDPC justifica su uso para el canal de datos (PDSCH/PUSCH) en bloques de transporte grandes? ¿Y qué ventaja de los códigos Polar justifica su uso para canales de control de longitud corta?

**(c)** Explica el rol del CRC exterior en los códigos CA-Polar (*CRC-Aided Polar*). ¿Cómo mejora el rendimiento del decodificador SCL respecto al SC básico?

**Solución P40 — LDPC vs Polar en 5G NR**  
*(Sesión 05, §5 — Comparación LDPC vs Polar y selección en 5G NR)*

**(a)** Asignación en 5G NR (3GPP TS 38.212):  
- **LDPC:** PDSCH (Physical Downlink Shared Channel), PUSCH (Physical Uplink Shared Channel) → canales de **datos** de usuario.  
- **Polar:** PDCCH (Physical Downlink Control Channel), PUCCH, PBCH (Physical Broadcast Channel), DCI, UCI → canales de **control** y señalización.

**(b)**  
**LDPC para bloques grandes:** la estructura quasi-cíclica permite decodicación en hardware altamente paralelo. A longitudes de bloque $n\geq1000$ bits, LDPC alcanza BER muy bajas con pocas iteraciones BP y excelente rendimiento a tasas altas ($r_c>0.5$).  
**Polar para bloques cortos (<512 bits):** demostrado teóricamente que son *capacity-achieving* para cualquier canal BMS. A longitudes cortas, CA-Polar+SCL supera a LDPC y turbo en BLER — justo lo que necesita el control (DCI de decenas de bits).

**(c)** En **CA-Polar** (*CRC-Aided*), se adjunta un CRC externo (11 o 24 bits) a los bits de información antes del encoding Polar. El decodificador SCL mantiene una **lista de $L$ paths** candidatos. Al terminar la decodificación, selecciona el path que satisface el CRC. Esto elimina paths erróneos que SC básico habría elegido incorrectamente. Resultado: ganancia de $\sim$1–2 dB en BLER vs SC a la misma $L$, y el error suelo (*error floor*) desaparece prácticamente.
